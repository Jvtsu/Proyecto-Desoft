# 📌 Flujo de trabajo en equipo con GitHub

Para evitar conflictos y mantener el repositorio ordenado, todos debemos seguir este flujo:

---

## 🔹 1. Siempre partir actualizados

Antes de trabajar:

```bash
git checkout main
git pull origin main
```

---

## 🔹 2. Crear una rama propia

Nadie trabaja directamente en `main`.

```bash
git checkout -b nombre-de-tu-rama
```

Ejemplos:

* `feature/fase`
* `fix/parsing`
* `grafico-derivadas`

---

## 🔹 3. Trabajar y guardar cambios

```bash
git add .
git commit -m "Descripción clara del cambio"
```

---

## 🔹 4. Subir tu rama

```bash
git push -u origin nombre-de-tu-rama
```

---

## 🔹 5. Crear Pull Request (PR)

En GitHub:

* Ir al repositorio
* Click en **"Compare & pull request"**
* Explicar qué hiciste

Otro compañero revisa antes de integrar.

---

## 🔹 6. Integrar cambios (merge)

Una vez aprobado el PR:

* Se hace merge a `main`
* El código queda actualizado para todos

---

## 🔹 7. Actualizar nuevamente

Después de cada merge:

```bash
git checkout main
git pull origin main
```

---

# ⚠️ Reglas importantes

* ❌ No trabajar directamente en `main`
* ❌ No hacer push directo a `main`
* ✔ Siempre usar ramas
* ✔ Siempre hacer pull antes de empezar
* ✔ Usar mensajes de commit claros

---

# 🚀 Flujo resumido

```bash
git checkout main
git pull

git checkout -b mi-rama
# trabajar
git add .
git commit -m "cambios"
git push -u origin mi-rama
```

Luego crear PR en GitHub.

---

# 🧠 Nota

Si hay conflictos:

```bash
git pull --rebase origin main
```

y resolver manualmente.

---

Este flujo evita errores, conflictos y pérdida de trabajo.
